```python
print("My Monthly Budget")
# Define the income
income=100000
# define the expence
food=20000
transport=12000
entertainment=8000
other=18000
# calculate saving
saving=(income-food-transport-entertainment-other)
print("Monthly saving :", saving)
```

```python
print("Monthly Budget Calculator")
#Define the income
income = float(input("Enter your monthly income: "))
#Define the expence
food = float(input("Enter your monthly food expenses: "))
transport = float(input("Enter your monthly transport expenses: "))
entertainment = float(input("Enter your monthly entertainment expenses: "))
other = float(input("Enter your other monthly expenses: "))
#Calculate the expenses
total_expenses = food + transport + entertainment + other
#Calculate the Saving
savings = income - total_expenses
# Print the Budget
print("\n--- Budget Summary ---")
print(f"Income:        ${income:.2f}")
print(f"Food:          ${food:.2f}")
print(f"Transport:     ${transport:.2f}")
print(f"Entertainment: ${entertainment:.2f}")
print(f"Other:         ${other:.2f}")
print(f"Total Expenses:${total_expenses:.2f}")
print(f"Savings:       ${savings:.2f}")

if savings >=0:
  print("You have saved this month !")
else:
 print("You have spent more than your income for this month !")
```
